# Titanic Demo for Survival Prediction

## Import Libraries and Data


```python
import joblib
import pandas as pd
```

```python
# I import my save data model to use with the demo program to predict survival of a new passenger.
model = joblib.load("Titanic_prediction_model.joblib")
# I also import the original data.
data = pd.read_csv("train.csv")
```

## Titanic Demo

```python
# I decided to make my demo program a menu. The menu has two options add a new passenger to predict the survival or lookup an existing passenger.
def titanic_demo():
    print("Welcome to the Titanic Survival Demo!")
    print("Choose an option:")
    print("1. Predict survival for a new passenger")
    print("2. Look up an existing passenger by ID")
    choice = input("Enter your choice (1 or 2): ")

    if choice == '1':
        # Prompt the user for input
        Pclass = int(input("Passenger Class: Enter 1 for 1st Class, 2 for 2nd Class, 3 for 3rd Class. "))
        Sex = int(input("Sex: Enter 1 for Male, 0 for Female. "))
        Age = float(input("Enter Age: "))
        SibSp = int(input("Enter how many number of spouses/siblings: "))
        Parch = int(input("Enter number of children/parents: "))
        Embarked = int(input("Enter 0 for Cherbourg, 1 for Queenstown, 2 for Southampton: "))

        # Create a DataFrame with the inputs
        data_update = pd.DataFrame([[Pclass, Sex, Age, SibSp, Parch, Embarked]], columns=['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Embarked'])

        # Make prediction using the loaded model
        prediction = model.predict(data_update)

        # Prints the new passengers details
        print("\nPassenger Details")
        print(f"Passenger Class: {data_update['Pclass'].iloc[0]}")
        print(f"Male = 1 and Female = 0. Sex: {data_update['Sex'].iloc[0]}")
        print(f"Age: {data_update['Age'].iloc[0]}")
        print(f"Number of Siblings/Spouses Aboard: {data_update['SibSp'].iloc[0]}")
        print(f"Number of Parents/Children Aboard: {data_update['Parch'].iloc[0]}")
        print(f"Cherbourg = 0, Queenstown = 1 and Southampton = 2. Embarked: {data_update['Embarked'].iloc[0]}")

        # Prints the prediction result
        if prediction[0] == 1:
            print("Prediction: Survived")
        else:
            print("Prediction: Not Survived")

    elif choice == '2':
        passenger_id = int(input("Enter Passenger ID: "))

        # Check if the passenger ID exists in the dataset
        if passenger_id in data['PassengerId'].values:
            passenger = data[data['PassengerId'] == passenger_id].iloc[0]

            # Display passenger details
            print("\nPassenger Details:")
            print(f"Passenger ID: {passenger['PassengerId']}")
            print(f"Name: {passenger['Name']}")
            print(f"Passenger Class: {passenger['Pclass']}")
            print(f"Sex: {passenger['Sex']}")
            print(f"Age: {passenger['Age']}")
            print(f"Number of Siblings/Spouses Aboard: {passenger['SibSp']}")
            print(f"Number of Parents/Children Aboard: {passenger['Parch']}")
            print(f"Embarked: {passenger['Embarked']}")

            # Check if the passenger survived
            survived = passenger['Survived']
            if survived == 1:
                print("The passenger survived.")
            else:
                print("The passenger did not survive.")
        else:
            print(f"No passenger found with ID: {passenger_id}")
    else:
        print("Invalid choice. Please enter 1 or 2.")

titanic_demo()
```
